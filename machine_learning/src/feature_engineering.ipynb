```python
import pandas as pd
import ast
import numpy as np
```

```python
INPUT_FILE = "../data/processed/cleaned_movies.csv"
OUTPUT_FILE = "../data/processed/final_dataset.csv"
```

```python
GENRES = [
    "Action",
    "Adventure",
    "Animation",
    "Comedy",
    "Crime",
    "Documentary",
    "Drama",
    "Family",
    "Fantasy",
    "History",
    "Horror",
    "Music",
    "Mystery",
    "Romance",
    "Science Fiction",
    "Thriller",
    "War",
    "Western"
]
```

```python
movies = pd.read_csv(INPUT_FILE)

print("Dataset loaded.")
print("Shape:", movies.shape)
```

```python
movies.head()
```

```python
movies["genres"] = movies["genres"].apply(
    lambda x: ast.literal_eval(x)
    if isinstance(x, str)
    else []
)

movies["production_countries"] = movies[
    "production_countries"
].apply(
    lambda x: ast.literal_eval(x)
    if isinstance(x, str)
    else []
)

print("List columns converted.")
```

```python
print(movies["genres"].head())
print(movies["production_countries"].head())
```

```python
for genre in GENRES:
    column_name = "genre_" + genre.lower().replace(" ", "_")

    movies[column_name] = movies["genres"].apply(
        lambda genres: 1 if genre in genres else 0
    )

print("Genre features created.")
```

```python
genre_columns = [
    "genre_" + genre.lower().replace(" ", "_")
    for genre in GENRES
]

movies[genre_columns].head()
```

```python
movies["main_country"] = movies[
    "production_countries"
].apply(
    lambda countries: countries[0]
    if len(countries) > 0
    else "Unknown"
)

print("Main country feature created.")
```

```python
movies[
    [
        "title",
        "production_countries",
        "main_country"
    ]
].head()
```

```python
movies["log_budget"] = np.log1p(
    movies["budget"]
)

print("Log budget feature created.")
```

```python
movies["log_revenue"] = np.log1p(
    movies["revenue"]
)

print("Log revenue feature created.")
```

```python
movies[
    [
        "budget",
        "log_budget",
        "revenue",
        "log_revenue"
    ]
].head()
```

```python
feature_columns = [
    "budget",
    "log_budget",
    "runtime",
    "release_year",
    "release_month",
    "original_language",
    "main_country"
]

genre_columns = [
    "genre_" + genre.lower().replace(" ", "_")
    for genre in GENRES
]

feature_columns.extend(genre_columns)

feature_columns.append("revenue")
feature_columns.append("log_revenue")

movies = movies[feature_columns]

print("Features selected.")
```

```python
print("Final dataset shape:", movies.shape)

print("\nFinal columns:")
print(movies.columns.tolist())
```

```python
movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Final dataset saved:")
print(OUTPUT_FILE)
```

```python
def get_director(crew):
    try:
        crew = ast.literal_eval(crew)

        for person in crew:
            if person.get("job") == "Director":
                return person.get("name")

    except:
        pass

    return "Unknown"
```

```python
credits["director"] = credits["crew"].apply(
    get_director
)
```

```python
credits[
    ["id", "director"]
].head(10)
```

```python
def get_main_cast(cast):

    try:
        cast = ast.literal_eval(cast)

        names = []

        for person in cast[:5]:
            if "name" in person:
                names.append(person["name"])

        return names

    except:
        return []
```

```python
credits["main_cast"] = credits["cast"].apply(
    get_main_cast
)
```

```python
def get_production_companies(companies):

    try:
        companies = ast.literal_eval(companies)

        names = []

        for company in companies[:5]:
            if "name" in company:
                names.append(company["name"])

        return names

    except:
        return []
```

```python
movies = pd.read_csv(
    "../data/processed/cleaned_movies.csv"
)

print(movies.shape)
print(movies.columns)
```

```python
credits_clean = credits[
    ["id", "director", "main_cast"]
].copy()
```

```python
companies_clean = raw_movies[
    ["id", "production_companies_clean"]
].copy()
```

```python
credits_clean["id"] = pd.to_numeric(
    credits_clean["id"],
    errors="coerce"
)

companies_clean["id"] = pd.to_numeric(
    companies_clean["id"],
    errors="coerce"
)

movies["id"] = pd.to_numeric(
    movies["id"],
    errors="coerce"
)
```

```python
movies = movies.merge(
    credits_clean,
    on="id",
    how="left"
)
```

```python
movies = movies.merge(
    companies_clean,
    on="id",
    how="left"
)
```
