```python
import joblib
import numpy as np
import pandas as pd

MODEL_FILE = "../models/elastic_net_model4.pkl"

model = joblib.load(MODEL_FILE)

print("Model loaded successfully.")
```

```python
def predict_revenue(
    budget,
    runtime,
    release_year,
    release_month,
    original_language,
    main_country,
    genres
):
    movie = {
        "budget": budget,
        "log_budget": np.log1p(budget),
        "runtime": runtime,
        "release_year": release_year,
        "release_month": release_month,
        "original_language": original_language,
        "main_country": main_country
    }

    for genre in [
        "action",
        "adventure",
        "animation",
        "comedy",
        "crime",
        "documentary",
        "drama",
        "family",
        "fantasy",
        "history",
        "horror",
        "music",
        "mystery",
        "romance",
        "science_fiction",
        "thriller",
        "war",
        "western"
    ]:
        movie["genre_" + genre] = (
            1 if genre.replace("_", " ").title() in genres else 0
        )

    movie_data = pd.DataFrame([movie])

    predicted_log_revenue = model.predict(movie_data)[0]

    predicted_revenue = np.expm1(
        predicted_log_revenue
    )

    return predicted_revenue
```

```python
predicted_revenue = predict_revenue(
    budget=165000000,
    runtime=148,
    release_year=2026,
    release_month=7,
    original_language="en",
    main_country="United States of America",
    genres=[
        "Action",
        "Adventure",
        "Science Fiction",
        "Thriller"
    ]
)

print(
    f"Predicted Box Office Revenue: "
    f"${predicted_revenue:,.2f}"
)
```

```python
result = {
    "predicted_revenue": round(predicted_revenue, 2),
    "predicted_revenue_millions": round(
        predicted_revenue / 1_000_000,
        2
    )
}

print("Prediction Result")
print("-----------------")
print(
    f"Predicted Revenue: "
    f"${result['predicted_revenue']:,.2f}"
)
print(
    f"Predicted Revenue: "
    f"${result['predicted_revenue_millions']:.2f} million"
)
```
